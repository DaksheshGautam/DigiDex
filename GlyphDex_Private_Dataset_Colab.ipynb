```
import json
from pathlib import Path
import numpy as np
from correction_data import prepare_exports
strategy = 'drawing'  # or 'writer' with at least three real writers
labels = json.loads(Path('models/characters/labels.json').read_text())
arrays, report = prepare_exports([Path('glyphdex-private-dataset.json')], labels, strategy)
np.savez_compressed('/content/website_corrections.npz', **arrays)
Path('/content/website_corrections.json').write_text(json.dumps(report, indent=2))
print({s:report[s] for s in ('train','validation','test')})
print(report['note'])
```
